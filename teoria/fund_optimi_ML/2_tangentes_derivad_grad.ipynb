# Tangentes, derivadas y gradientes
Todos son conceptos bastante importantes en el machine learning.

Abordemos el cáculo de la pendiente entre dos puntos de una función dada. 

![pendiente](../../imagenes/pendiente.png)

se halla la pendiente que hay entre la línea roja que se crea al unir la evaluación de los puntos $x_k$ y $x_{k+1}$ y por tanto su evaluación en la función sus valores serán $y_k$ y $y_{k+1}$ y la distancia entre los puntos serán $\Delta x$ y  $\Delta y$ respectivamente. 

\begin{split} m & = \frac{y_{k+1} - y_k}{x_{k+1} - x_k} \\ \\ & = \frac{f(x_{k+1}) - f(x_k)}{x_{k+1} - x_k} \end{split}

Si se tiene en cuenta que $\Delta x = x_{k+1} - x_k y x_{k+1} = x_k + \Delta x$, entonces:

\begin{split} m & = \frac{f(x_{k+1}) - f(x_k)}{x_{k+1} - x_k} \\ \\ & = \frac{f(x_k + \Delta x) - f(x_k)}{\Delta x} \\ \end{split}

Recordemos que el concepto de derivada es el límite de cuando \Delta tiene a Cero lo cual viene a representar que la distancia entre $x_k$ y $x_{k+1}$ es cada vez más pequeña y finalmente queda una recta tangente en el punto $x_k$ y por ende la derivada será esta pendiente indicandome si la función es creciente o decreciente en aquel punto específico. 

![derivada](../../imagenes/derivada.png)

$$\frac{d}{dx} f(x_k) = \lim_{\Delta x \to 0} \frac{f(x_k + \Delta x) - f(x_k)}{\Delta x}$$

![signo](../../imagenes/signo.png)


## Derivada y puntos de máxima y mínima
![concavidad](../../imagenes/concavidad.png)

Los máximos y mínimos son lugares donde la primera derivada es cero, todo depende del comportamiento de cada función pues exáctamente donde estos puntos se hacen cero es entonces donde tendremos un máximo o mínimo. 


## Algunas derivadas comunes y útiles a tener en cuenta. 

### Derivada de una constante (línea horizontal)
En este caso la derivada es cero, ya que todos los valores de C son lo mismo. 

### Derivada de una recta. 
Para una rectra de la forma $y = ax + b$, su derivada es $a$.

### Derivada de una cuadrática
Para una función cuadrática, $y = x^2$ que es un parábola, su derivada es $y' = 2x$, que es una recta de pendiente 2 

Nota: De forma general la derivada de $x^n$ es $nx^{n-1}$. Para $1/x$ se toma es $x{-1}$


### Derivada de la inversa de una función
Función:

$$y = f(x)$$

Función inversa:

$$g(y) = x$$

Por definición:

$$g(f(x)) = x$$

Derivada:

$$g'(y) = \frac{1}{f'(x)}$$

### Derivada de funciones trigonométricas
$$f(x) = \sin(x), f'(x) = \cos(x)$$

$$f(x) = \cos(x), f'(x) = - \sin(x)$$

### Derivada de la función exponencial y logarítmica.
$$\frac{d}{dx} \exp (x) = \exp(x)$$

$$\frac{d}{dx} \log(x) = \frac{1}{x} $$

## Propiedades más comunes de las derivadas

### Derivada  por una constante
\begin{split} \frac{d}{dx} f(x) & = \frac{d}{dx} C \cdot g(x) \\ \\ & = C \cdot \frac{d}{dx} g(x) \\ \\ \end{split}

### Derivada de una suma
$$\frac{d}{dx} \left[ f(x) + g(x) \right] = \frac{d}{dx} f(x) + \frac{d}{dx} g(x)$$

### Derivada del producto
$$\frac{d}{dx} \left[ f(x) \cdot g(x) \right] = \left( f(x) \frac{d}{dx} g(x) \right) + \left( g(x) \frac{d}{dx} f(x) \right)$$

## Derivadasd parciales y gradiente
Recordemos que la derivada para una función de dos variables, se basa en usar la otra variable como constante. 

\begin{split} f_x & = \frac{\partial}{\partial x} f(x, y) \\ \\ & = \lim_{\Delta x \to 0} \frac{f(x + \Delta x, y) - f(x, y)}{\Delta x} \\ \end{split}

\begin{split} f_y & = \frac{\partial}{\partial y} f(x, y) \\ \\ & = \lim_{\Delta y \to 0} \frac{f(x, y + \Delta y) - f(x, y)}{\Delta y} \\ \end{split}

## Vector gradiente
Se define como el vector de las derivadas parciale de una función: 

$$\nabla f(x, y) = \left[ \begin{split} \frac{\partial}{\partial x} f(x, y) \\ \\ \frac{\partial}{\partial y} f(x, y) \\ \end{split} \right]$$

* La dirección del vector gradiente indica donde la función aumenta más rápido. (apunta la dirección máxima de crecimiento)

* La máxima tasa de cambio: la magnitud del vector es igual al valor de la derivada direccional en ese punto.

* Si el vector gradiente apunta a la dirección donde se maximiza la función entonces el - vector gradiente apunta a la dirección donde se minimiza la función.

## Método del gradiente descendente
Es un proceso iterativo en que el que el valor actual de x se corrige sumandole un valor $\Delta x$ con signo contrario a la derivada con el fin de alcanzar el punto de mínima.

La ecuación de corrección o mejora es:

$$x_k = x_{k-1} - \mu \frac{d}{dx} f(x_{k-1})$$

donde $\mu$ es la tasa o factor de aprendizaje, $x_k$ el siguiente punto a calcular, $x_{k-1}$ el punto conocido. El signo de menos hace que se busque el mínimo que permite el mejoramiento continuo. 

![gradiente](../../imagenes/gradiente.png)

Por tanto, en esta metología se parte de un punto $x_0$ el cuál permite calcular $x_1$ se tal manera que se minimiza un poco el valor de la función en el punto, con ello se halla $x_2$, $x_3$ y así sucesivamente hasta encontrar el valor de x donde la función se minimiza. 

Para funciones de variables variables: 

$$\mathbf{w}_k = \mathbf{w}_{k-1} - \mu \frac{\partial}{\partial \mathbf{w}} f(\mathbf{w}_{k-1})$$

con $\mathbf{w}_k$ un vector columna. 

### Ejemplo del descenso por gradiente con la función de Rosenbrock.
$$f(x, y) = 100 \cdot (x^2 - y)^2 + (1 - x)^2$$

para:

$$ x \in [-2.048, 2.048]$$

$$ y \in [-1.000, 4.000]$$

y

$$f(1, 1) = 0$$

en (1,1) es que se tiene ese punto de mínima y allí la función vale 0.

Derivadas parciales: 
\begin{split} \frac{\partial}{\partial x} f(x, y) & = \frac{\partial}{\partial x} \left[ 100 (x^2 - y)^2 + (1-x)^2 \right] \\ \\ & = \frac{\partial}{\partial x} \left[ 100 (x^2 - y)^2 \right] + \frac{\partial}{\partial x} \left[ (1-x)^2 \right] \\ \\ & = 100 \cdot 2 \cdot (x^2 - y) \frac{\partial}{\partial x} (x^2 - y) + 2 \cdot (1 - x)\frac{\partial}{\partial x} (1-x) \\ \\ & = 100 \cdot 2 \cdot (x^2 - y) \frac{\partial}{\partial x} (x^2) + 2 \cdot (1 - x)\frac{\partial}{\partial x} (-x) \\ \\ & = 200 \cdot (x^2 - y) \cdot 2x + 2 \cdot (1 - x) \cdot (-1) \\ \\ & = 400 x (x^2 - y) - 2 (1- x) \end{split}


\begin{split} \frac{\partial}{\partial y} f(x, y) & = \frac{\partial}{\partial y} \left[ 100 (x^2 - y)^2 + (1-x)^2 \right] \\ \\ & = \frac{\partial}{\partial y} \left[ 100 (x^2 - y)^2 \right] + \frac{\partial}{\partial y} \left[ (1-x)^2 \right] \\ \\ & = 100 \cdot 2 \cdot (x^2 - y) \frac{\partial}{\partial y} (x^2 - y) + 2 \cdot (1 - x)\frac{\partial}{\partial y} (1-x) \\ \\ & = -200 \cdot (x^2 - y) \frac{\partial}{\partial y} (-y) + 2 \cdot (1 - x) \cdot 0 \\ \\ & = -200 \cdot (x^2 - y) \end{split}


Por tanto, el vector gradiente es: 

$$g(x, y) = \frac{\partial}{\partial \mathbf{w}} f(\mathbf{w}) = \left[ \begin{array}{c} 400 x (x^2-y) - 2(1 - x) \\ -200(x^2 - y) \end{array} \right]$$



```python
#
# Función
#
def f(x, y):
    return 100 * (x ** 2 - y) ** 2 + (1 - x) ** 2


#
# Gradiente
#
def g(x, y):
    grad_x = 400 * x * (x ** 2 - y) - 2 * (1 - x)
    grad_y = -200 * (x ** 2 - y)
    return grad_x, grad_y


#
# Función de mejora
#
def improve(x, y, mu):
    grad_x, grad_y = g(x, y)
    x = x - mu * grad_x
    y = y - mu * grad_y
    return x, y


#
# Punto de inicio
#
x = -0.5
y = +3.5

history = {"x": [x], "y": [y], "f": [f(x, y)]}


#estas dos líneas son el corazón del método
for epoch in range(50):
    x, y = improve(x, y, mu=0.001)

    history["x"].append(x)
    history["y"].append(y)
    history["f"].append(f(x, y))

#
# Ultimo resultado obtenido
#
print("      x = {:6.5f}\n      y = {:+6.5f}\nf(x, y) = {:+6.5f}".format(x, y, f(x, y)))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import cm


def plot_contour():

    #definir números para X
    X = np.linspace(start=-2.048, stop=2.048, num=50)
    #definir número para Y
    Y = np.linspace(start=-0.1, stop=4.0, num=50)
    #combinaciones para evaluzar funcion
    X, Y = np.meshgrid(X, Y)

    #función
    Z = f(X, Y)

    #contour es para gráficas curvas de nivel, contourf es para gráficas de nivel rellenas
    plt.subplots(figsize=(4, 4))
    plt.gca().contourf(X, Y, Z, cmap=cm.Greys, levels=20, alpha=1.0)
    plt.gca().contour(X, Y, Z, colors="gray", levels=20)

    # el punto de mínima es el que aparece como un círculo con un punto negro
    plt.plot(
        [1], [1], "o", color="black", fillstyle="none", markersize=11, markeredgewidth=2
    )
    plt.plot([1], [1], ".", color="black")

    plt.xlabel("x")
    plt.ylabel("y")


plot_contour()
#ir gráficando paso a paso con puntos negros rellenos los pasos
plt.plot(history["x"], history["y"], "o-", color="k")
plt.show()
```
