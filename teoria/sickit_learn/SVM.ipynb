# Máquina de sorporte vectorial (SVM)

La Máquina de Vectores de Soporte (Support Vector Machine / SVM), cuando se aplica a regresión (Support Vector Regression / SVR), cambia el paradigma tradicional del Error Cuadrático Medio (MSE).

En lugar de intentar minimizar la distancia de todos los puntos a la recta, SVR construye un tubo de tolerancia de ancho $2\epsilon$ alrededor del plano de predicción: los puntos que caen dentro del tubo son ignorados (tienen error cero), y el modelo solo se preocupa por penalizar aquellos que quedan fuera.

1. El Tubo de Margen $\epsilon$-insensibleEn una regresión lineal estándar, cualquier desviación por pequeña que sea genera una penalización cuadrática. En SVR se define una función de pérdida $\epsilon$-insensible ($L_\epsilon$:

$$L_\epsilon(y, f(\mathbf{x})) = \max(0, \vert{}y - f(\mathbf{x})\vert{} - \epsilon)$$

* Dentro del tubo ($\vert{}y - f(\mathbf{x})\vert{} \le \epsilon$): La pérdida es exactamente 0.

* Fuera del tubo ($\vert{}y - f(\mathbf{x})\vert{} > \epsilon$): La pérdida crece linealmente según la distancia al borde del tubo.

2. Variables de Holgura ($\xi_i, \xi_i^*$) y la Función Objetivo

Como en la práctica es difícil contener todos los puntos dentro del tubo sin hacer el modelo inflexible, introducimos variables de holgura ($\xi_i$ para desviaciones por encima del tubo y $\xi_i^*$ por debajo).

El problema se formula como una optimización cuadrática con restricciones:

$$\min_{\mathbf{w}, b, \xi, \xi^*} \frac{1}{2} \Vert{}\mathbf{w}\Vert{}^2 + C \sum_{i=1}^{N} (\xi_i + \xi_i^*)$$

Sujeto a:

$$\begin{cases}  y_i - (\mathbf{w}^T \mathbf{x}_i + b) \le \epsilon + \xi_i \\  (\mathbf{w}^T \mathbf{x}_i + b) - y_i \le \epsilon + \xi_i^* \\  \xi_i, \xi_i^* \ge 0  \end{cases}$$

¿Qué significan estos términos?

* $\frac{1}{2} \Vert{}\mathbf{w}\Vert{}^2$ (Regularización / Suavidad): Minimizar la norma del vector de pesos $\mathbf{w}$ equivale a aplanar la función para evitar oscilaciones bruscas (previene el overfitting).

* Hiperparámetro $C$: Controla el balance entre mantener la función plana y permitir violaciones al tubo. Un $C$ grande penaliza severamente los puntos fuera del tubo; un $C$ pequeño tolera más errores a cambio de una curva más suave.

3. La Formulación Dual y los Vectores de SoporteAplicando los Multiplicadores de Lagrange ($\alpha_i, \alpha_i^* \ge 0$), el problema se convierte en su forma Dual:

$$f(\mathbf{x}) = \sum_{i=1}^{N} (\alpha_i - \alpha_i^*) \mathbf{x}_i^T \mathbf{x} + b$$

Aquí surge la propiedad fundamental del algoritmo: 

* Para todos los puntos que caen estrictamente dentro del tubo, las condiciones de KKT (Karush-Kuhn-Tucker) fuerzan a que $\alpha_i = 0$ y $\alpha_i^* = 0$.

* Únicamente los puntos que están sobre el borde o fuera del tubo tienen multiplicadores $\alpha_i \ne 0$.

Vectores de Soporte: Son exclusivamente aquellos datos cuyos coeficientes $\alpha_i \ne 0$. El modelo final depende únicamente de este subconjunto de puntos críticos; el resto del dataset puede eliminarse sin alterar la función aprendida.

4. El "Truco del Kernel" (Kernel Trick)Si la relación entre $X$ y $y$ no es lineal, sustituimos el producto escalar directo $\mathbf{x}_i^T \mathbf{x}_j$ por una función Kernel $K(\mathbf{x}_i, \mathbf{x}_j)$:$$f(\mathbf{x}) = \sum_{i \in \text{SV}} (\alpha_i - \alpha_i^*) K(\mathbf{x}_i, \mathbf{x}) + b$$El Teorema de Mercer garantiza que $K(\mathbf{x}_i, \mathbf{x}_j) = \phi(\mathbf{x}_i)^T \phi(\mathbf{x}_j)$, lo que significa que estamos mapeando los datos a un espacio de características de dimensión superior ($\phi$) y calculando el producto escalar allí, sin evaluar ni almacenar explícitamente las coordenadas proyectadas.
