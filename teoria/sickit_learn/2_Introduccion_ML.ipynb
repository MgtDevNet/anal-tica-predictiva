## Conversión entre tipos de datos
Es otro elemento clave y fundamental en la creación de nuestros modelos es la conversión de tipos de datos, el hecho de cuidar el espacio en memoria para las variables y todo esto, nos permite que nuestro proyecto sea más liviano. 

```python
#
# Cuando sea posible, una entrada de tipo 'float32' se mantendra.
# De otra forma, se hara la conversión a float64.
#
import numpy as np
from sklearn import kernel_approximation

rng = np.random.RandomState(0) # generar semilla para aleatoriedad, 
#este es un generador. 
X = rng.rand(10, 2000) # generar matriz de 10 x 2000
X = np.array(X, dtype="float32") # convertirlo en array de numpy tipo float32
display(X.dtype)
```

```python
transformer = kernel_approximation.RBFSampler()
X_new = transformer.fit_transform(X)
X_new.dtype
```

El código anterior sirve para aproximar la transformación del "Truco del Kernel RBF" (Radial Basis Function / Kernel Gaussiano) de forma explícita, convirtiendo tus datos originales a un espacio de mayor dimensión sin el alto costo computacional del algoritmo SVR estándar.

`transformer = kernel_approximation.RBFSampler()`
 Crea un objeto generador de características aleatorias de Fourier (Random Fourier Features).

El Kernel RBF (Gaussiano) proyecta tus datos a una dimensión infinita para encontrar patrones no lineales complejos. Como una matriz de dimensión infinita no se puede calcular en una computadora real, RBFSampler utiliza un teorema matemático (Teorema de Bochner) para crear un número finito de características (por defecto 100) que imitan y aproximan esa dimensión superior.

`X_new = transformer.fit_transform(X)`  Toma tu matriz de características original X (de tamaño $N \text{ filas} \times D \text{ columnas}$) y la transforma en una nueva matriz X_new (de tamaño $N \text{ filas} \times 100 \text{ columnas}$).

.fit(X): Analiza la escala y número de dimensiones del dataset $X$..transform(X): Aplica proyecciones de funciones seno y coseno con frecuencias aleatorias a cada dato.Resultado: Mapea una variable $x$ no lineal a un nuevo espacio donde una simple Regresión Lineal convencional puede trazar un plano recto y obtener un rendimiento casi idéntico al de un modelo no lineal complejo (como SVR(kernel='rbf')).

```python
#
# Algunos estimadores hacen la conversión a float64 por estabilidad numérica
#
from sklearn import datasets
from sklearn.svm import SVC

iris = datasets.load_iris()
clf = SVC() # creación de la instancia del clasificador
clf.fit(iris.data, iris.target) # ajsute y entrenamiento

# Los targets en modelos de regresión son convertiddos
# a float64
list(
    clf.predict(iris.data[:3]), # Predicciones para los primero 3 datos. 
)
```

```python
clf.fit(
    iris.data,
    iris.target_names[iris.target],
)

# El tipo de los targets se mantiene en modelos de clasificación
list(
    clf.predict(iris.data[:3]),
)
```

## Re-ajsute y modificación de parámetros


```python
#
# Es posible reajustar y modificar los valores de los parámetros de un
# estimador
#
import numpy as np
from sklearn.datasets import load_iris
from sklearn.svm import SVC

X, y = load_iris(return_X_y=True)

# No se especifican los parámetros del estimador
clf = SVC() # instancia de máquina de vectores de soporte de clasificación. 

# Se ajusta el estimador usando 'set_params'
clf.set_params(kernel="linear").fit(X, y) # para cambiar los parámetros
# cambiando a que el kernel sea lienal. 
clf.predict(X[:5])
```

```python
# Se ajusta el estimador usando 'set_params'
clf.set_params(kernel="rbf").fit(X, y)
clf.predict(X[:5])
```

Decir que un kernel es lineal significa, en términos simples, que no estás haciendo ninguna transformación mágica a los datos; los estás dejando exactamente en su espacio original.

La intuición: Buscar patrones en el mundo realImagínate que tus datos son puntos en una hoja de papel ($2D$):Usar un Kernel Lineal: Intentas separar los datos o trazar la tendencia dibujando una línea recta directamente sobre el papel con una regla. Si usas $3D$, es un plano plano.Usar un Kernel NO Lineal (como el RBF Gaussiano): Si los datos forman un círculo o una curva compleja, no los puedes separar con una regla recta. Lo que hace un Kernel No Lineal es "doblar el papel" o proyectar los puntos a un espacio $3D$ (como si los catapultaras hacia arriba a diferentes alturas) para que, en esa tercera dimensión, sí puedas pasar una hoja plana metálica que los divida limpiamente.

Matemáticamente, todos los kernels calculan una medida de similitud entre dos puntos $x_1$ y $x_2$.El Kernel Lineal es simplemente el Producto Escalar tradicional:
$$\text{Kernel}_{\text{lineal}}(x_1, x_2) = x_1 \cdot x_2$$

Mide qué tan alineados están dos vectores en una línea recta. Si los datos ya son separables en su forma original, esto basta.

El Kernel No Lineal transforma antes de multiplicar:

$$\text{Kernel}_{\text{no-lineal}}(x_1, x_2) = \phi(x_1) \cdot \phi(x_2)$$

Aplica una función $\phi$ que convierte los puntos a curvas, parábolas o infinitas dimensiones antes de medir su similitud.

## Multiclase vs Multietiqueta
Sickit learn también permite trabajar con clasificadores multiclase o multi-etiqueta.  

Un clasificador multietiqueta (Multi-label Classifier) es un modelo de aprendizaje supervisado en el que una misma observación puede asignarse a varias categorías o etiquetas al mismo tiempo.

A diferencia de la clasificación tradicional (donde cada ejemplo pertenece a una y solo una categoría), aquí las etiquetas no son mutuamente excluyentes.

Ejemplo para entender la diferencia
Clasificación tradicional (Multi-clase): Le muestras una foto de un animal al modelo y debe responder si es Perro, Gato o Pájaro (solo una opción).

Clasificación Multietiqueta: Le muestras una película al modelo y este le asigna los géneros: ["Acción", "Ciencia Ficción", "Aventura"] (múltiples opciones simultáneas).

casos múy usados en clasificación de texto/noticias y etiquetado de imágenes. 


```python
#
# Ajuste multiclase vs muultietiqueta
#
from sklearn.multiclass import OneVsRestClassifier
from sklearn.preprocessing import LabelBinarizer
from sklearn.svm import SVC

X = [  # Ejemplos
    [1, 2],  # 1
    [2, 4],  # 2
    [4, 5],  # 3
    [3, 2],  # 4
    [3, 1],  # 5
]

# Clases: {0, 1, 2}
y = [
    0,
    0,
    1,
    1,
    2,
]

#
# Clasificación multiclase
# Este el tipo de casos o modelso que se usando cuando se 
# trata de diferentes tipos de clases. Este modelo es conocido
# como ** Uno contra el resto **. El elemento se le asigna a la clase
# que más aparece en la salida. 
classif = OneVsRestClassifier(
    estimator=SVC(random_state=0), # instancia del clasificador. 
)
classif.fit(X, y).predict(X)
```

```python
#
# Clasificación multietiqueta
# Para casos que elementos pueden tener diferentes etiquetas. 
y = LabelBinarizer().fit_transform(y) # transformador de la 
# varaible objetivo, toma las clases y las vuelve etiquetas que indican
# la pertenencia a una clase o no. 
y
#       0  1  2  <--- clases
# ---------------------------
# y direcamente la pertenencia de un registros la clase
# es con un 1 y la no pertenencia con un 0 y hay tantas columnas
# como clases. La primera columna a la clase 0, la segunda a la 1 y así. 
```

```python
classif.fit(X, y).predict(X)  # para los ultimos 2 ejemplos no se asigna clase
# esto es lo peligros de este tipo de clasificadores pues pueden predecir
# que no hay predicción para algunos casos

# Por tanto, en la predicción hay tantas flas como observaciones y tantas
# columnas como clases, para también lo que se predice que una clase puede pertenecer. 
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

# Es posible asignar más de una clase a un ejemplo.
# Las etiquetas son 0, 1, 2, 3, y 4.
y = [
    [0, 1],
    [0, 2],
    [1, 3],
    [0, 2, 3],
    [2, 4],
]

y = MultiLabelBinarizer().fit_transform(y)
y
#       0  1  2  3  4 <--- etiquetas
# ------------------------------------

#por tanto a los X de entrada se les quiere asignar 1, 2, 3 o 4
# de las posibles etiquetas. Como resultado, hay tantas filas
# como observaciones y tantas columnas como clases. 
```

```python
classif.fit(X, y).predict(X)
#       0  1  2  3  4 <--- etiquetas
# -------------------------------------
```

# Introducción al aprendizaje estadístico
Es la disciplina que busca es construir modelos basado en estadística que permiten modelar un conjunto de pruebas determinado ya sea con modelos de regresión o clasificación. 

```python
#
# Ejemplo de un dataset disponible en sklearn
#
from sklearn import datasets

iris = datasets.load_iris()
data = iris.data

# el formato es (n_samples, n_features). si no tiene este formato
# debe ser preprocesado
data.shape
```

```python
# este data set contiene ejemplos de las imagenes de 8x8 de los dígitos del 0 al 9
# donde las X son las matrices que representan las imágenes y el Y
# directamente al número al que pertenecen. 
digits = datasets.load_digits()
digits.images.shape
import matplotlib.pyplot as plt

plt.imshow(
    digits.images[-1],
    cmap=plt.cm.gray_r, # imagen de la matriz en la esacala de grises. 
)
```

```python
# Se necesita transformar la matriz de 8x8 que represente el ejemplo
# y por ende hacemos un reshape para que quede todo como un solo vector fila 
#y por obvias razones queda como un vector de 64 columnas. 
data = digits.images.reshape((digits.images.shape[0], -1))
data.shape
```

## Estimadores. 
(Recordemos que los estimadores en sickit learn es cualquier objeto, o algoritmo que aprende a partir de los datos. )

```python
#
# Parámetros y estimación
# 
from sklearn import linear_model

#instancia de un modelo de regresión lineal y decimos que se ajuste también
# por mínimos cuadrados el intercepto. 
estimator = linear_model.LinearRegression(fit_intercept=True)

# Nota: El parámetro fit_intercept controla si el modelo calcula o
#  no el término independiente $\beta_0$ (también conocido como
#  intersección, bias o sesgo) en la ecuación de la recta/plano. 
# Tiene la interpretación geométrica de cuando todos los valores
# de las predictoras son cero. Solo se recomienda poner False
# si se esta seguro de que las predictoras nunca serán cero todas 
# simultáneamente.

estimator.fit(data, digits.target) # X , Y acá se hace la estimación del modelo
# es decir, el cálculo de los estimadores para las varaibles
# por mínimos cuadrados ordinarios. 
```

```python
#
# Parámetros estimados
# Ya que se hizo el ajuste, entocnes acá se tiene  las estimaciones de las 
# 64 columnas (variables) de nuestro set de datos. 
estimator.coef_
```
