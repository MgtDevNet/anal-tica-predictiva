## Visualización de funciones en 3d
La visualización de funciones es particularmente importante en la parte de optimización no lineal, específicamente cuando se están abordando las técnicas numéricas para enconctrar el mínimo o máximo de una función. Para esto se usa la librería matplotlib. 

Para esto, se arealiza un ejemplo con la función **de Rosenbrock** de dos dimensiones que se define como: 

$$f(x, y) = 100 \cdot (x^2 - y)^2 + (1 - x)^2$$
para:

$$x \in [-2.048, 2.048]$$

$$y \in [-1.000, 4.000]$$

y

$$f(1,1)=0$$

```python
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return 100 * (x**2 - y) ** 2 + (1 - x) ** 2
```

El proceso de visualización consta de 3 pasos. 

1. Definir una malla fina en el plano X-Y. 
2. Calcular la función $z = f(x,y)$ para cada punto en el plano. 
3. visualizar a z. 

```python
import numpy as np

#
# Generación de un vector de puntos para x
# 10 puntos entre -2.048 y 2.048
X = np.linspace(start=-2.048, stop=2.048, num=10)
X.round(3) #redondear a 3 decimales

#
# Generación de un vector de puntos para y
#
Y = np.linspace(start=-0.1, stop=4.0, num=10)
Y.round(3)
```

```python
import numpy as np

# 1. Define 1D coordinate vectors
a = np.array([1, 2, 3])
b = np.array([10, 20])

# 2. Generate the 2D grid
a, b = np.meshgrid(a, b)

print("a matrix:\n", a) # tiene b filas de valores
print("\nb matrix:\n", b) # tiene a columnas de valores
```

```python
#
# Generación de una matriz para representar
# una malla de puntos
#
X, Y = np.meshgrid(X, Y)

X.round(3) # X con tantas filas como elementos tiene Y y tantas columnas como elementos tiene X

#Explicitamente np.meshgrid convierte vectores de coordenadas unidimensionales en matrices de coordenadas bidimensionales.
# o multidimensionales para evaluar funciones en una cuadríacula. 
# lo cuál se hace para realizar calculos de funciones que dependen de dos variables
# y con ellos se puede realizar un gráfico en 3 dimensiones. 

#esto se hace para poder evaluar la función f(x,y) en todos los puntos de la malla generada por X y Y.
```

```python
Y.round(3) # Y con tantas filas como elementos tiene X y tantas columnas como elementos tiene Y
```

Si X tiene 10 valores y Y tiene 10 valores, la malla será 10 x 10 pues serán todas las combinaciones posibles de puntos.

```python
# Ahora creamos un diagarama de dispersión con estos puntos
# de tal manera uqe crean la malla de puntos que serían
# todas las combinaciones de los valores de X y Y.
plt.figure(figsize=(8,6))
plt.scatter(X, Y, s = 10)
plt.show()
```

Ahora evaluamos la función z = f(x,y) que habíamos construido desde antes. 

```python
#
# Generación de una matriz con los valores de f(x, y)
#
Z = f(X, Y)
Z.round(1) # será una matriz 10 x 10
#type(Z) # es un array (matriz de numpy)
#Z.shape # es una matriz 10 x 10
```

para graficas una función en 3 dimensiones se usa el parámetros projection = "3d" y esta función recibe como parámetros  el azim (el ángulo de rotación horizontal de la cámara al ahacer renderizaods) y la elevación 

```python
fig = plt.figure(figsize=(10, 7))

i = 1
for elev in [10, 60]: # 2 elevaciones
    for azim in [-60, 0, 60]: # 3 ángulos de azimut

        ax = fig.add_subplot(2, 3, i, projection="3d", azim=azim, elev=elev)
        i += 1
        plt.xlabel("x")
        plt.ylabel("y")
        plt.title("elev={}, azim={}".format(elev, azim))

plt.tight_layout()
plt.show()
```

Ahora bien, ya con lo anterior tenemos nuestra malla, nuestro espacio para gráficar. Por tanto, en matplotlib contamos con la función plot_surface() para gráficos en 3 dimensiones. 

```python
#
# Gráfica de la función
#
from matplotlib import cm # importando colormaps

X = np.linspace(start=-2.048, stop=2.048, num=50)
Y = np.linspace(start=-0.1, stop=4.0, num=50)
X, Y = np.meshgrid(X, Y)
Z = f(X, Y)

ax = plt.figure(figsize=(4, 4)).add_subplot(projection="3d")

# plot_suface traza la superficie conectando todos los puntos. 
ax.plot_surface(
    X,
    Y,
    Z,
    cmap=cm.binary, # es simplemente un mapa de colores que se aplica a la superficie
    linewidth=1, # grososr de líneas de la malla 
    antialiased=False # desactiva suavizado de líneas
)

# notese que los puntos negros son los qaue representan los puntos más altos de 
#la función 

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Otra alternativa es la función plot_wireframe() que es una malla de alambres que construye la figura

```python
# Primero se crea el tamaño de la figura y se añade la proyección 3d
ax = plt.figure(figsize=(4, 4)).add_subplot(projection="3d")

ax.plot_wireframe(
    X,
    Y,
    Z,
    color="black",
    linewidth=0.8,
    alpha=1.0, #transparencia
    rstride=2, # distancia entre filas, cada 2 puntos
    cstride=2, # distancia entre columnas, cada 2 puntos. 
)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Aún así no es muy claro la visualización de la función, por ellos es que pueden convinarse los dos métodos anteriores. plot_wireframe() y plot.surface() para mejorar la visualización.  

```python
ax = plt.figure(figsize=(4, 4)).add_subplot(projection="3d")

ax.plot_surface(
    X,
    Y,
    Z,
    cmap=cm.binary,
    linewidth=1,
    antialiased=False,
)

ax.plot_wireframe(
    X,
    Y,
    Z,
    color="black",
    linewidth=0.8,
    alpha=1.0,
    rstride=2,
    cstride=2,
)

plt.xlabel("x")
plt.ylabel("y")
plt.show()

# litertalmente basta con dar el tamaño e indicar la proyección 3d, 
# de resto simplemente hay se dibuja una gráfica sobre la otra. 
```

También es posible dibujar fácilamente una gráfica de contorno o curva de nivel  con la función plot.contour()

```python
#
# Líneas de nivel con el mismo color
#
fig = plt.figure(figsize=(4, 4)) #tamaño de la figura

plt.gca().contour( # llamar función 
    X, # matriz eje X
    Y, # matriz eje Y
    Z, # matriz eje Z
    colors="gray",
    levels=20, # número de niveles a ver. 
)

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
#
# Líneas de nivel con diferentes colores
#
fig = plt.figure(figsize=(3, 3))
plt.gca().contour(
    X, # matriz eje X
    Y, # matriz eje Y
    Z, # matriz eje Z
    cmap=cm.Greys, # colormap para las líneas de nivel
    # poner un mapa de color es la clave para que se usen
    # los colores del colormap en lugar de un solo color.
    # a más oscura la línea, más alto es el valor de la función en ese punto.
    levels=20,
)

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

también la función plt.contourf() está para mostrar las curvas de nivel de una manera diferente con respecto a los valores donde la función es mayor. 

```python
fig = plt.figure(figsize=(3, 3))

plt.gca().contourf(
    X, # matriz eje X
    Y, # matriz eje Y
    Z, # matriz eje Z
    cmap=cm.Greys,
    levels=20,
)

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

también es posible y recomendabo usar contour() y contourf() para mejorar la visualización

```python
fig = plt.figure(figsize=(3, 3))

plt.gca().contourf(
    X,
    Y,
    Z,
    cmap=cm.Greys,
    levels=20,
    alpha=1.0,
)

plt.gca().contour(
    X,
    Y,
    Z,
    colors="gray",
    levels=20,
)


plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Para gráficar el púnto mínimo, hacemos el gráfico y tambié se grafica un solo punto (1,1) pues ahí valoe cero. 




```python
fig = plt.figure(figsize=(3, 3)) #  tamaño de la figura

plt.gca().contourf( # función contourf
    X,
    Y,
    Z,
    cmap=cm.Greys,
    levels=20,
    alpha=1.0,
)

plt.gca().contour( # función contour
    X,
    Y,
    Z,
    colors="gray",
    levels=20,
)

plt.plot( # grafica un punto en el mínimo de la función
    [1], # 1 en X
    [1], # 1 en Y
    "o", # símbolo del punto y será en todo ese centro
    color="black", #negro 
    fillstyle="none", 
    markersize=11,
    markeredgewidth=2,
)

plt.plot(
    [1], # 1 en X
    [1], # 1 en Y
    ".", # símbolo del punto
    color="black",
)

plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
